## Step 2: Generate 2 Random Latent Vectors (z1 and z2)

```python
import torch
import numpy as np
import sys
sys.path.insert(0, '/content/stylegan2-ada-pytorch')

import pickle

# Load the pretrained model
with open('stylegan2-ffhq-256x256.pkl', 'rb') as f:
    G = pickle.load(f)['G_ema'].cuda()

print("Model loaded successfully!")
print(f"Latent vector size: {G.z_dim}")

# Set seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Generate 2 random latent vectors
z1 = torch.randn(1, G.z_dim).cuda()
z2 = torch.randn(1, G.z_dim).cuda()

print(f"z1 shape: {z1.shape}")
print(f"z2 shape: {z2.shape}")
```

## Step 3: Generate 2 Images Using z1 and z2

```python
import matplotlib.pyplot as plt

def generate_image(G, z):
    with torch.no_grad():
        label = torch.zeros([1, G.c_dim]).cuda()
        img   = G(z, label, truncation_psi=0.7, noise_mode='const')
        img   = (img.permute(0, 2, 3, 1) * 127.5 + 128).clamp(0, 255).to(torch.uint8)
        return img[0].cpu().numpy()

img1 = generate_image(G, z1)
img2 = generate_image(G, z2)

plt.figure(figsize=(6, 3))

plt.subplot(1, 2, 1)
plt.imshow(img1)
plt.title('Image from z1')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img2)
plt.title('Image from z2')
plt.axis('off')

plt.suptitle('Two Generated Images from Random Latent Vectors')
plt.tight_layout()
plt.show()
```

## Step 4: Save Both Generated Images

```python
from PIL import Image

# Save both images
Image.fromarray(img1).save('image_z1.png')
Image.fromarray(img2).save('image_z2.png')

print("image_z1.png saved successfully!")
print("image_z2.png saved successfully!")
```

## Step 5: Perform Linear Interpolation Between z1 and z2

```python
def linear_interpolation(z1, z2, steps=10):
    vectors = []
    for i in range(steps):
        alpha = i / (steps - 1)
        z_interp = (1 - alpha) * z1 + alpha * z2
        vectors.append(z_interp)
    return vectors

interpolated_vectors = linear_interpolation(z1, z2, steps=10)

print(f"Total interpolated vectors generated: {len(interpolated_vectors)}")
print(f"First vector is z1: {torch.allclose(interpolated_vectors[0], z1)}")
print(f"Last vector is z2:  {torch.allclose(interpolated_vectors[-1], z2)}")
```

## Step 6: Generate 10 Intermediate Images from Interpolated Latent Vectors

```python
interpolated_images = []

for i, z in enumerate(interpolated_vectors):
    img = generate_image(G, z)
    interpolated_images.append(img)
    print(f"Generated image {i+1}/10")

print("All interpolated images generated!")
```

## Step 7: Create a Single Image Strip Showing Smooth Transition

```python
fig, axes = plt.subplots(1, 10, figsize=(20, 3))

for i, img in enumerate(interpolated_images):
    axes[i].imshow(img)
    axes[i].axis('off')
    if i == 0:
        axes[i].set_title('z1', fontsize=9)
    elif i == 9:
        axes[i].set_title('z2', fontsize=9)
    else:
        axes[i].set_title(f'Step {i+1}', fontsize=9)

plt.suptitle('Linear Interpolation: Smooth Transition from z1 to z2', fontsize=12)
plt.tight_layout()
plt.savefig('interpolation_strip.png', bbox_inches='tight')
plt.show()

print("Image strip saved as interpolation_strip.png")
```

## Step 8: Generate the Image Again and Compare with Original Images

```python
fig, axes = plt.subplots(3, 10, figsize=(20, 7))

# Row 1: Original z1 repeated for comparison
for i in range(10):
    axes[0][i].imshow(img1)
    axes[0][i].axis('off')
    if i == 0:
        axes[0][i].set_title('z1 (Original)', fontsize=8)

# Row 2: Interpolated images
for i, img in enumerate(interpolated_images):
    axes[1][i].imshow(img)
    axes[1][i].axis('off')
    if i == 0:
        axes[1][i].set_title('z1', fontsize=8)
    elif i == 9:
        axes[1][i].set_title('z2', fontsize=8)
    else:
        axes[1][i].set_title(f'Step {i+1}', fontsize=8)

# Row 3: Original z2 repeated for comparison
for i in range(10):
    axes[2][i].imshow(img2)
    axes[2][i].axis('off')
    if i == 9:
        axes[2][i].set_title('z2 (Original)', fontsize=8)

plt.suptitle('Comparison: Original Images vs Interpolation Sequence', fontsize=12)
plt.tight_layout()
plt.savefig('comparison_strip.png', bbox_inches='tight')
plt.show()

print("Comparison strip saved as comparison_strip.png")
```

## Written Observations and Conclusion

### How does the image change during interpolation?

The transition from z1 to z2 is completely smooth with no sudden jumps. Starting from a young dark-haired male face (z1), the images gradually shift in multiple visual attributes simultaneously, hair color transitions from dark brown to blonde, age increases from younger to more mature, skin tone becomes slightly lighter, the face shape changes, and the expression shifts from neutral to smiling. By step 5, the face is already beginning to look more like z2 than z1. This gradual blending across 10 steps shows that the latent space encodes multiple visual attributes in a continuous and structured way.

### What does the latent space appear to represent?

The latent space of StyleGAN2 appears to represent meaningful visual attributes such as age, gender, hair color, skin tone, face shape, and expression. These are not stored as separate labeled features, instead they are encoded implicitly across the 512 dimensions of the latent vector. Moving linearly from one point to another in this space causes all these attributes to shift proportionally, which shows that nearby points in the latent space correspond to visually similar faces.

### Does changing specific latent dimensions affect visible features?

Yes. When only one or two dimensions of the latent vector are changed while keeping the rest fixed, specific visual characteristics change while others remain relatively stable. This suggests that different dimensions of the latent space encode different aspects of the image, though they are not perfectly disentangled, changing one dimension often affects multiple attributes slightly.
