# Assignment 2: Exploring Text Generation Using GPT-2

**Objective:** To understand how different text-generation parameters
influence the output produced by a language model.

```python
!pip install transformers torch
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

model = GPT2LMHeadModel.from_pretrained("gpt2")
```

## Opening Sentence (used for all 5 experiments)

The following prompt is kept the same across all experiments. Only one parameter is changed each time to observe how it affects the output.

```python
prompt = "The future of technology depends on the choices we make today."
```

```python
# Converting into tokens
inputs = tokenizer(prompt, return_tensors="pt")
```

```python
output = model.generate(
    **inputs,
    max_new_tokens=50
)
```

```python
# Decode it next
# Removes hidden control tokens-like padding, beginning-of-sequence, or end-of-sequence markers when converting token IDs back into readable text
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

---
## Experiment 1 — Low Temperature (0.3)
**Parameter changed:** temperature = 0.3

Low temperature makes the model very confident. It sticks to the most probable tokens, so the output is more predictable and repetitive.

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.3
)
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

---
## Experiment 2 — High Temperature (1.5)
**Parameter changed:** temperature = 1.5

High temperature flattens the probability distribution. The model picks more unusual tokens, making the output more creative but sometimes random or incoherent.

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.5
)
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

---
## Experiment 3 — Small Top-k (top_k=5)
**Parameter changed:** top_k = 5

With top_k=5, only the top 5 most probable tokens are considered at each step. This restricts the vocabulary heavily and leads to safe but predictable output.

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    top_k=5
)
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

---
## Experiment 4 — Low Top-p (top_p=0.5)
**Parameter changed:** top_p = 0.5

Top-p (nucleus sampling) limits the model to the smallest set of tokens whose cumulative probability reaches 0.5. At 0.5, only the very top tokens are used — output tends to be focused but less varied.

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    top_k=0,
    top_p=0.5
)
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

---
## Experiment 5 — High Repetition Penalty (2.0)
**Parameter changed:** repetition_penalty = 2.0

A high repetition penalty strongly discourages the model from reusing words. This forces more variety in the output, but sometimes at the cost of natural flow.

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.0,
    repetition_penalty=2.0
)
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print(generated_text)
```

## Custom Prompt: Three Generations

**Prompt type:** Story opening

> *"She opened the old wooden door and found a letter that changed everything."*

```python
prompt = "She opened the old wooden door and found a letter that changed everything."

inputs = tokenizer(prompt, return_tensors="pt")
```

### Custom Generation 1: Balanced settings
temperature=0.7, top_k=50, top_p=0.95, repetition_penalty=1.2

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.7,
    top_k=50,
    top_p=0.95,
    repetition_penalty=1.2
)

generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```

### Custom Generation 2: More creative
temperature=1.2, top_k=0, top_p=0.95, repetition_penalty=1.0

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=1.2,
    top_k=0,
    top_p=0.95,
    repetition_penalty=1.0
)

generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```

### Custom Generation 3: Controlled and clean
temperature=0.5, top_k=40, top_p=1.0, repetition_penalty=1.5

```python
output = model.generate(
    **inputs,
    max_new_tokens=80,
    do_sample=True,
    temperature=0.5,
    top_k=40,
    top_p=1.0,
    repetition_penalty=1.5,
    pad_token_id=tokenizer.eos_token_id
)

generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```

---
## Brief Analysis

**Temperature** had the most noticeable effect across all experiments. At 0.3, the output was fluent but repetitive, the model kept returning to similar phrases. At 1.5, it became much more unpredictable, sometimes switching topics mid-sentence or using words that did not quite fit the context. This matches the theory: temperature controls how "sharp" the probability distribution is, and extreme values push the model toward either repetition or chaos.

**Top-k** at 5 was very restrictive. The model produced sensible text, but it felt narrow, like it was working with a limited vocabulary. Increasing top-k generally allows more variety without making the output incoherent.

**Top-p** at 0.5 had a similar effect to a low top-k, keeping the output focused. The advantage of top-p over top-k is that it adapts to context, at uncertain steps, it allows more candidates; at confident steps, it restricts to fewer. This tends to feel more natural.

**Repetition penalty** at 2.0 clearly reduced loops in the output. Without it, GPT-2 tends to repeat the same phrase over and over (as seen in the demo notebook output). At 2.0, the model actively avoided reusing tokens, which produced more varied text, but occasionally the grammar suffered because the model was avoiding even necessary repeated words.

**Additional Activity:** At temperature=1.5, the output can become meaningless or jump between unrelated ideas. This happens because the probability distribution becomes nearly flat, every token gets a similar chance of being picked, so the model stops following its learned patterns of what logically comes next. The output looks like text but lacks coherent meaning.
